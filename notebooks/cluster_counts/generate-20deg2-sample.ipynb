# Generate Inputs for Example Pipeline

This notebook generates mock shear and photometry catalogs for a $\sim 20\,\mathrm{deg}^2$ patch in CosmoDC2. Then we load redmapper clusters from the same healpix pixels. These are the inputs needed to run the example pipeline.

```python
import sys
import os
import ceci
# get the TXPipe root directory (two levels up from this notebook).
# notebook_dir is only captured on the first run, so re-running this
# cell after the chdir below doesn't move us further up the tree.
if "notebook_dir" not in globals():
    notebook_dir = os.path.abspath("")
txpipe_dir = os.path.abspath(os.path.join(notebook_dir, "../.."))
sys.path.append(txpipe_dir)
# relative paths below (e.g. ./data/...) are relative to the TXPipe root
os.chdir(txpipe_dir)
import txpipe
```

```python
# These six pixels come to about 23 deg^2
pixels = [8786, 8787, 8788, 8914, 8915, 9043]
outdir = "./data/cosmodc2/inputs-20deg2"
extra_cols="redshift_true"
response_model=f"{txpipe_dir}/data/DESY1-R-model.hdf5"

# prepare to run
os.makedirs(outdir, exist_ok=True)

# configure the stage
stage = txpipe.TXCosmoDC2Mock.make_stage(
    response_model=response_model,
    shear_catalog=f"{outdir}/shear_catalog.hdf5",
    photometry_catalog=f"{outdir}/photometry_catalog.hdf5",
    healpixels=pixels,
    extra_cols=extra_cols
)
```

```python
# run the stage
stage.run()

# move the stage outputs to their final locations
stage.finalize()
```

```python
# make the matching redMaPPer cluster catalog, cut to the same healpixels

import GCRCatalogs
import healpy as hp
import h5py
import numpy as np

# cosmoDC2 healpix_pixels are nside=32, RING ordering
nside = 32
cols = ["ra", "dec", "richness", "richness_err", "cluster_id",
        "scaleval", "redshift", "redshift_err"]

cl_cat = GCRCatalogs.load_catalog("cosmoDC2_v1.1.4_redmapper_v0.8.1")
cl_data = cl_cat.get_quantities(cols)

cl_pix = hp.ang2pix(nside, cl_data["ra"], cl_data["dec"], lonlat=True)
keep = np.isin(cl_pix, pixels)
print(f"Keeping {keep.sum()} of {keep.size} clusters in pixels {pixels}")

with h5py.File(f"{outdir}/cluster_catalog.hdf5", "w") as f:
    g = f.create_group("clusters")
    for name, col in cl_data.items():
        g.create_dataset(name, data=col[keep])
```
