## WL sample selection

Our first step is the WL sample selection, made by three stages: `TXMetaDetectSelection` applies the S/N, size and flag cuts (`shear_selection`), `TXSimpleRedshiftSourceTomography` puts the objects in tomographic bins of true redshift (`shear_bin_assignments`), and `TXComputeShearCalibrationMetadetect` combines them and measures the calibration (`shear_tomography_catalog`).

````python
out = "data/test/metadetect/outputs_metadetect"

step1 = txpipe.TXMetaDetectSelection.make_stage(
    # This file is the input metadetect shear catalog
    shear_catalog="data/example/inputs/metadetect_shear_catalog.hdf5",
    # This is the output file for this stage: whether each object passes the cuts
    shear_selection=f"{out}/shear_selection.hdf5",
    output_dir=out,
)
# tomographic bins from the true redshifts
step1b = txpipe.TXSimpleRedshiftSourceTomography.make_stage(
    shear_catalog="data/example/inputs/metadetect_shear_catalog.hdf5",
    shear_bin_assignments=f"{out}/shear_bin_assignments.hdf5",
    output_dir=out,
    redshift_col="redshift_true",
    source_zbin_edges=[0.1, 0.5, 1.0, 3.0],
    chunk_rows=100_000,
)
# calibration of the selected objects
step1c = txpipe.TXComputeShearCalibrationMetadetect.make_stage(
    shear_catalog="data/example/inputs/metadetect_shear_catalog.hdf5",
    shear_bin_assignments=f"{out}/shear_bin_assignments.hdf5",
    shear_selection=f"{out}/shear_selection.hdf5",
    shear_tomography_catalog=f"{out}/shear_tomography_catalog.hdf5",
    output_dir=out,
    delta_gamma=0.02,
)
````

We run them in order:

````python
for step in [step1, step1b, step1c]:
    step.run()
    step.finalize()
````

## Cluster-source pairs

Our second step runs the matching to find the shear catalog behind every cluster.

This step saves only the cluster-source pairs, which is actually just an index into the shear and cluster catalogs (to avoid making many copies of the data). The weights and shears are computed by the ensemble stage.

````python
print("Options for this pipeline and their defaults:")
print(txpipe.extensions.CLClusterSourceSelect.config_options)

step2 = txpipe.extensions.CLClusterSourceSelect.make_stage(
    # Shear catalog, as before
    shear_catalog="data/example/inputs/metadetect_shear_catalog.hdf5",
    # This is the initial cluster catalog - RAs, Decs, richess, redshift, etc.
    cluster_catalog="./data/example/inputs/cluster_catalog.hdf5",
    # This fiducial cosmology is used to convert distance separations to redshifts
    fiducial_cosmology="./data/fiducial_cosmology.yml",
    # The selection and tomography catalogs created in step 1 select objects for the WL sample.
    # By default the sources must pass the cuts and be in one of the tomographic bins.
    shear_selection=f"{out}/shear_selection.hdf5",
    shear_tomography_catalog=f"{out}/shear_tomography_catalog.hdf5",
    # This is a QP file created by RAIL to generate the photo-zs for this sample
    source_photoz_pdfs="data/example/inputs/photoz_pdfs.hdf5",

    # This is the output for this stage
    cluster_shear_selection="my_cluster_shear_selection.hdf5",

    # Let's override one of the configuration parameters for this stage:
    max_radius=5.0
)
````

````python
step2.run()
step2.finalize()
````

## Exploring the sources of each cluster

To avoid making lots and lots of copies of the data, this stage has not made a catalog, but instead made an index into the other catalogs.

The ensemble stage computes the weights, the calibrated shears and the tangential and cross components of the sources of the clusters in its bins, and saves them in `cluster_source_quantities`. We have a helper class which is designed to match up the cluster catalog, these sources and the profiles, and collect the results for each cluster.

````python
# The shears, weights and profiles are computed by the ensemble stage, which needs
# the clusters in bins. It saves the sources of each cluster with these quantities
# in cluster_source_quantities, which the helper class below reads.
step_bin = txpipe.extensions.CLClusterBinningRedshiftRichness.make_stage(
    cluster_catalog="./data/example/inputs/cluster_catalog.hdf5",
    cluster_catalog_tomography="my_cluster_catalog_tomography.hdf5",
    richedge=[1, 80],
    zedge=[0.2, 1.0],
)
step_bin.run()
step_bin.finalize()

step_ens = txpipe.extensions.CLClusterEnsembleProfiles.make_stage(
    shear_catalog="data/example/inputs/metadetect_shear_catalog.hdf5",
    shear_tomography_catalog=f"{out}/shear_tomography_catalog.hdf5",
    fiducial_cosmology="./data/fiducial_cosmology.yml",
    cluster_shear_selection="my_cluster_shear_selection.hdf5",
    cluster_catalog_tomography="my_cluster_catalog_tomography.hdf5",
    source_photoz_pdfs="data/example/inputs/photoz_pdfs.hdf5",
    cluster_profiles="my_cluster_profiles.pkl",
    cluster_source_quantities="my_cluster_source_quantities.hdf5",
)
step_ens.run()
step_ens.finalize()
````

````python
ccc = txpipe.extensions.CombinedClusterCatalog(
    cluster_catalog="./data/example/inputs/cluster_catalog.hdf5",
    cluster_source_quantities="my_cluster_source_quantities.hdf5",
    cluster_profiles="my_cluster_profiles.pkl",
    shear_catalog="data/example/inputs/metadetect_shear_catalog.hdf5",
)
````

## Investigating our pipeline

While that's running, let's have a look at what's in the pipeline.

- The pipeline file on IN2P3 is `examples/mocks/cosmodc2/Cluster_pipelines/pipeline-20deg2-CL-in2p3.yml`
- The pipeline file on NERSC is `examples/mocks/cosmodc2/Cluster_pipelines/pipeline-20deg2-CL-nersc.yml`

Comment/uncomment out the appropriate `pipeline_file` definition below, depending on whether you are running on IN2P3 or on NERSC

First, we can use ceci to build a flow-chart showing the pipeline stages:

````python
# Read the appropriate pipeline configuration, and ask for a flow-chart.
# pipeline_file = "examples/mocks/cosmodc2/Cluster_pipelines/pipeline-20deg2-CL-in2p3.yml"
pipeline_file = "examples/mocks/cosmodc2/Cluster_pipelines/pipeline-20deg2-CL-nersc.yml"
flowchart_file = "20deg2.png"


pipeline_config = ceci.Pipeline.build_config(
    pipeline_file,
    flow_chart=flowchart_file,
    dry_run=True
)

# Run the flow-chart pipeline
ceci.run_pipeline(pipeline_config)
````

Now we can have a look at the chart it has created:

````python
Image(flowchart_file)
````

The flowchart elements are classified as follows:

- Red ellipses are pipeline stages, each of which is a python class.

- Yellow boxes are pre-existing input data files.

- Blue boxes are files created by the pipeline.

We can say that it generally makes sense - the final outputs are the cluster profiles and the sources of each cluster, just like before.

Let's have a look at the pipeline information for this stage:


````python
pprint(pipeline_config)
````

This dictionary defines what pipeline stages are run, and how they are executed. You can see:

- a list of stages to be run, including their parallelization.
- site information showing how to run individual steps.
- directories to put logs and outputs
- launcher information on how to launch and manage the workflow
- overall inputs to the pipeline

Finally, the 'config' item points to another file that configures the individual pipeline stages:



````python
with open(pipeline_config['config']) as f:
    print(f.read())
````
